## Data Cleaning and Missing Values Visualization

```python
sns.heatmap(df.isnull(),cbar=False)
```

```python
df.isnull().sum()
```

```python
df = df.drop('Cabin', axis=1,errors="ignore")
display(df.head())
```

```python

df["FamilySize"] = df["SibSp"] + df["Parch"] +1
df.tail(5)
```

```python
df["IsAlone"] = (df["FamilySize"]==1).astype(int)
df.tail(5)
```

```python
df["Title"] = df["Name"].str.extract(
    r',\s*([^.]*)\.',
    expand=False
)
df["Title"] = df["Title"].replace(
    ["Lady", "Countess", "Capt", "Col", "Don", "Dr",
     "Major", "Rev", "Sir", "Jonkheer", "Dona"],
    "Rare"
)

df["Title"].replace(
    ["Dr", "Rev", "Sir"],
    "Rare"
)
```

```python
df["FareByPerson"] = df["Fare"]/df["FamilySize"]
df.head(5)
```

```python
df["Sex_Pclass"] = (df["Sex"]+"_"+df["Pclass"].astype(str))
df.head(5)
```

```python
df = df.drop(["PassengerId","Name","Ticket"],axis=1,errors='ignore')
df.head(5)
```

```python
x = df.drop("Survived",axis=1)
y = df["Survived"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
num_features = ["Age","Fare","SibSp","Parch","FamilySize","Pclass","IsAlone","FareByPerson"]
cat_features = ["Embarked","Sex","Sex_Pclass","Title"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
num_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
    ])
cat_pipeline = Pipeline([
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("encoder",OneHotEncoder(handle_unknown="ignore",drop="first"))
    ])
```

```python
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer([
    ("num",num_pipeline,num_features),
    ("cat",cat_pipeline,cat_features)
])
```

```python
x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed = preprocessor.transform(x_test)
```

```python
df.columns
```

```python
import numpy as np
np.isnan(x_train_processed).sum()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
model = Pipeline([
    ("preprocessor",preprocessor),
    ("model",LogisticRegression(random_state=42))
])
model.fit(x_train,y_train)
y_predict = model.predict(x_test)
accuracy = accuracy_score(y_test,y_predict)
print(accuracy)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_predict)
cm
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
ax = sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)
ax.set_xlabel("Predicted")
ax.set_ylabel("Actual")
ax.set_title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report
classification_report(y_test,y_predict)

```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
dt_model = DecisionTreeClassifier(max_depth=3,random_state=42)
dt_model.fit(x_train_processed,y_train)
dt_predict = dt_model.predict(x_test_processed)
accuracy = accuracy_score(y_test,dt_predict)
print("Accuracy of Decision Tree:")
print(accuracy)
print("Classification report of Decision Tree before tuning:")
print(classification_report(y_test,dt_predict))
print("Confusion Matrix of Decision Tree before tuning:")
print(confusion_matrix(y_test,dt_predict))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
dt_model =Pipeline([
    ("preprocessor",preprocessor),
    ("model",DecisionTreeClassifier(random_state=42))
])
dt_param_grid = {
    "model__max_depth": [3, 5, 7, 10, None],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4],
    "model__criterion": ["gini", "entropy"]
}
dt_grid_search = GridSearchCV(
    dt_model,
    dt_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
dt_grid_search.fit(x_train,y_train)
dt_tuned_predict = dt_grid_search.predict(x_test)
dt_tuned_accuracy = accuracy_score(y_test,dt_tuned_predict)
print("Accuracy of Decision Tree after tuning:")
print(dt_tuned_accuracy)
print("Best parameters:")
print(dt_grid_search.best_params_)
print("Best cv score")
print(dt_grid_search.best_score_)
print("Classification report of Decision Tree after tuning:")
print(classification_report(y_test,dt_tuned_predict))
print("Confusion Matrix of Decision Tree after tuning:")
print(confusion_matrix(y_test,dt_tuned_predict))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
rf_model =RandomForestClassifier(n_estimators=100,max_depth=3,random_state=42)
rf_model.fit(x_train_processed,y_train)
rf_predict = rf_model.predict(x_test_processed)
accuracy = accuracy_score(y_test,rf_predict)
print("Accuracy of Random Forest:")
print(accuracy)
print("Classification report of Random Forest before tuning:")
print(classification_report(y_test,rf_predict))
print("Confusion Matrix of Random Forest before tuning:")
print(confusion_matrix(y_test,rf_predict))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = Pipeline([
    ("preprocessor",preprocessor),
    ("model",RandomForestClassifier(random_state=42))
])

rf_param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [3, 5, 7],
    "model__min_samples_split": [2, 5],
    "model__min_samples_leaf": [1, 2]
}
rf_grid_search = GridSearchCV(
    rf_model,
    rf_param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)


rf_grid_search.fit(x_train,y_train)
rf_tuned_predict = rf_grid_search.predict(x_test)
rf_tuned_accuracy = accuracy_score(y_test,rf_tuned_predict)
print("Accuracy of Random Forest after tuning:")
print(rf_tuned_accuracy)
print("Best parameters:")
print(rf_grid_search.best_params_)
print("Best cv score:")
print(rf_grid_search.best_score_)
print("Classification report of Random Forest after tuning:")
print(classification_report(y_test,rf_tuned_predict))
print("Confusion Matrix of Random Forest after tuning:")
print(confusion_matrix(y_test,rf_tuned_predict))
```
